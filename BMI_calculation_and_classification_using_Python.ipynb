**BMI Calculator**


```python
patient_name = "alyana"
age = 40
print(patient_name)
print(age)
```

```python
# Input patient data
height_meters = 1.79
weight_kg = 71.2
print(height_meters)
print(weight_kg)
```

```python
# Calculate BMI
bmi = weight_kg / (height_meters ** 2)
```

```python
#Display
print(f"\nAlyana BMI: {bmi:.2f}")
```

```python
# BMI Classification on Conditions
if bmi < 18.5:
    print("Alyana is underweight.")
elif 18.5 <= bmi < 24.9:
    print("Alyana has a normal weight.")
elif 25 <= bmi < 29.9:
    print("Alyana is overweight.")
else:
    print("Alyana has obesity.")
```

# Result:
Alyana’s BMI is 22.22 kg/m², which falls within the normal BMI range (18.5–24.9 kg/m²) based on her height and weight.

```python
import matplotlib.pyplot as plt
```

```python
# BMI categories
categories = ['Underweight', 'Normal', 'Overweight', 'Obese']
# Upper limit of each category
bmi_values = [18.5, 24.9, 29.9, 40]
# Alyana's BMI
alyana_bmi = 22.22
```

```python
colors = ['lightblue', 'lightgreen', 'orange', 'red']
plt.bar(categories, bmi_values, color=colors)

# Highlight Alyana's BMI with a dot
plt.plot(1, alyana_bmi, marker='o', markersize=12, color='blue', label=f"Alyana's BMI: {alyana_bmi}")

plt.ylabel('BMI Value')
plt.title('BMI Categories with Alyana\'s BMI')
```
